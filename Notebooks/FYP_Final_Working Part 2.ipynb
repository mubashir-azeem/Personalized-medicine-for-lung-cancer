```python
import torch
from torch_geometric.data import Data
from rdkit import Chem

def atom_features(atom):

    return [
        atom.GetAtomicNum(),
        atom.GetDegree(),
        atom.GetFormalCharge(),
        int(atom.GetIsAromatic())
    ]

def smiles_to_graph(smiles):

    mol = Chem.MolFromSmiles(smiles)

    if mol is None:
        return None

    x = []

    for atom in mol.GetAtoms():
        x.append(atom_features(atom))

    x = torch.tensor(
        x,
        dtype=torch.float
    )

    edge_index = []

    for bond in mol.GetBonds():

        start = bond.GetBeginAtomIdx()
        end = bond.GetEndAtomIdx()

        edge_index.append([start,end])
        edge_index.append([end,start])

    edge_index = torch.tensor(
        edge_index,
        dtype=torch.long
    ).t().contiguous()

    return Data(
        x=x,
        edge_index=edge_index
    )


drug_graphs = {}

for _, row in smiles_df.iterrows():

    graph = smiles_to_graph(
        row["CanonicalSMILES"]
    )

    if graph is not None:

        drug_graphs[
            row["Drug"]
        ] = graph

print("Total Graphs:", len(drug_graphs))
```

```python
import torch.nn as nn
import torch.nn.functional as F

from torch_geometric.nn import (
    GCNConv,
    global_mean_pool
)

class DrugGNN(nn.Module):

    def __init__(self):

        super().__init__()

        self.conv1 = GCNConv(
            4,
            64
        )

        self.conv2 = GCNConv(
            64,
            128
        )

        self.conv3 = GCNConv(
            128,
            256
        )

    def forward(self,data):

        x = data.x
        edge_index = data.edge_index

        batch = torch.zeros(
            x.size(0),
            dtype=torch.long,
            device=x.device
        )

        x = self.conv1(
            x,
            edge_index
        )

        x = F.relu(x)

        x = self.conv2(
            x,
            edge_index
        )

        x = F.relu(x)

        x = self.conv3(
            x,
            edge_index
        )

        x = global_mean_pool(
            x,
            batch
        )

        return x
```

```python
device = torch.device("cpu")

drug_gnn = DrugGNN().to(device)

drug_gnn.eval()

drug_embeddings = {}

for drug, graph in drug_graphs.items():

    graph = graph.to(device)

    with torch.no_grad():

        emb = drug_gnn(graph)

    drug_embeddings[drug] = (
        emb.cpu()
           .numpy()
           .flatten()
    )

print(
    "Drug Embeddings Created:",
    len(drug_embeddings)
)
```

```python
drug_emb_df = pd.DataFrame.from_dict(
    drug_embeddings,
    orient="index"
)

drug_emb_df.reset_index(
    inplace=True
)

drug_emb_df.rename(
    columns={
        "index":"DRUG_NAME"
    },
    inplace=True
)

print(drug_emb_df.shape)

display(
    drug_emb_df.head()
)
```

```python
# NO ATTENTION MODEL:

from tensorflow.keras.layers import (
    Input,
    Dense,
    Concatenate,
    Dropout
)

from tensorflow.keras.models import Model

cell_input = Input(
    shape=(256,),
    name="CancerEmbedding"
)

drug_input = Input(
    shape=(256,),
    name="DrugEmbedding"
)

merged = Concatenate()(
    [
        cell_input,
        drug_input
    ]
)

x = Dense(
    512,
    activation="relu"
)(merged)

x = Dropout(0.3)(x)

x = Dense(
    256,
    activation="relu"
)(x)

x = Dropout(0.3)(x)

x = Dense(
    128,
    activation="relu"
)(x)

output = Dense(
    1,
    name="LN_IC50"
)(x)

no_attention_model = Model(
    inputs=[
        cell_input,
        drug_input
    ],
    outputs=output
)

no_attention_model.compile(
    optimizer="adam",
    loss="mse",
    metrics=["mae"]
)

no_attention_model.summary()
```

```python
cell_ic50 = (
    training_df
    .groupby("SangerModelID_x")["LN_IC50"]
    .mean()
    .reset_index()
)

print(cell_ic50.shape)
cell_ic50.head()
```

```python
cell_mapping = training_df[
    ["ModelID", "SangerModelID_x"]
].drop_duplicates()

print(cell_mapping.shape)
cell_mapping.head()
```

```python
# Expression Data
NSCLC_expression_scaled = pd.read_csv('/content/NSCLC_expression_scaled.csv')
expr = NSCLC_expression_scaled.copy()

# Merge Cell Mapping

expr = expr.merge(
    cell_mapping,
    on="ModelID",
    how="inner"
)

# Merge Mean IC50

task_df = expr.merge(
    cell_ic50,
    on="SangerModelID_x",
    how="inner"
)

print(task_df.shape)

task_df.head()
```

```python
import numpy as np

gene_cols = NSCLC_expression_scaled.columns[1:]

X_gene = task_df[gene_cols].values

y_ic50 = task_df["LN_IC50"].values

print(X_gene.shape)
print(y_ic50.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_gene,
    y_ic50,
    test_size=0.2,
    random_state=42
)

print(X_train.shape)
print(X_test.shape)
```

```python
from tensorflow.keras.layers import (
    Input,
    Dense,
    BatchNormalization,
    Dropout
)

from tensorflow.keras.models import Model
from tensorflow.keras.optimizers import Adam

input_layer = Input(
    shape=(8146,),
    name="GeneInput"
)

# Encoder

x = Dense(
    2048,
    activation="relu"
)(input_layer)

x = BatchNormalization()(x)

x = Dropout(0.3)(x)

x = Dense(
    1024,
    activation="relu"
)(x)

x = BatchNormalization()(x)

x = Dropout(0.3)(x)

x = Dense(
    512,
    activation="relu"
)(x)

latent = Dense(
    256,
    activation="relu",
    name="LatentEmbedding"
)(x)

# Decoder

d = Dense(
    512,
    activation="relu"
)(latent)

d = Dense(
    1024,
    activation="relu"
)(d)

d = Dense(
    2048,
    activation="relu"
)(d)

reconstruction = Dense(
    8146,
    name="Reconstruction"
)(d)

# IC50 Prediction Head

p = Dense(
    128,
    activation="relu"
)(latent)

p = Dense(
    64,
    activation="relu"
)(p)

ic50_output = Dense(
    1,
    name="IC50Prediction"
)(p)

task_autoencoder = Model(
    inputs=input_layer,
    outputs=[
        reconstruction,
        ic50_output
    ]
)

task_autoencoder.compile(
    optimizer=Adam(1e-4),
    loss={
        "Reconstruction":"mse",
        "IC50Prediction":"mse"
    },
    loss_weights={
        "Reconstruction":1.0,
        "IC50Prediction":5.0
    }
)

task_autoencoder.summary()
```

```python
from tensorflow.keras.callbacks import EarlyStopping

early_stop = EarlyStopping(
    monitor="val_loss",
    patience=15,
    restore_best_weights=True
)

history = task_autoencoder.fit(
    X_train,
    {
        "Reconstruction":X_train,
        "IC50Prediction":y_train
    },
    validation_data=(
        X_test,
        {
            "Reconstruction":X_test,
            "IC50Prediction":y_test
        }
    ),
    epochs=200,
    batch_size=16,
    callbacks=[early_stop],
    verbose=1
)
```

```python
from sklearn.model_selection import train_test_split

(
    X_cell_train,
    X_cell_test,
    X_drug_train,
    X_drug_test,
    y_train,
    y_test
) = train_test_split(
    X_cell,
    X_drug,
    y,
    test_size=0.2,
    random_state=42
)
```

```python
from tensorflow.keras.layers import (
    Input,
    Dense,
    Dropout,
    Concatenate,
    Reshape,
    Flatten,
    MultiHeadAttention,
    LayerNormalization
)

from tensorflow.keras.models import Model
from tensorflow.keras.optimizers import Adam

cell_input = Input(
    shape=(272,),
    name="CellInput"
)

cell_proj = Dense(
    256,
    activation="relu"
)(cell_input)

drug_input = Input(
    shape=(256,),
    name="DrugInput"
)

cell_seq = Reshape((1,256))(cell_proj)
drug_seq = Reshape((1,256))(drug_input)

attn1 = MultiHeadAttention(
    num_heads=4,
    key_dim=64
)(
    cell_seq,
    drug_seq
)

attn1 = LayerNormalization()(attn1)

attn2 = MultiHeadAttention(
    num_heads=4,
    key_dim=64
)(
    drug_seq,
    cell_seq
)

attn2 = LayerNormalization()(attn2)

attn1 = Flatten()(attn1)
attn2 = Flatten()(attn2)

merged = Concatenate()([
    cell_proj,
    drug_input,
    attn1,
    attn2
])

x = Dense(
    512,
    activation="relu"
)(merged)

x = Dropout(0.3)(x)

x = Dense(
    256,
    activation="relu"
)(x)

x = Dropout(0.3)(x)

x = Dense(
    128,
    activation="relu"
)(x)

output = Dense(
    1,
    name="LN_IC50"
)(x)

final_attention_model = Model(
    inputs=[
        cell_input,
        drug_input
    ],
    outputs=output
)

final_attention_model.compile(
    optimizer=Adam(
        learning_rate=1e-4
    ),
    loss="mse",
    metrics=["mae"]
)

final_attention_model.summary()
```

```python
from tensorflow.keras.callbacks import EarlyStopping

early_stop = EarlyStopping(
    monitor="val_loss",
    patience=10,
    restore_best_weights=True
)

history = final_attention_model.fit(
    [
        X_cell_train,
        X_drug_train
    ],
    y_train,
    validation_split=0.2,
    epochs=100,
    batch_size=128,
    callbacks=[early_stop],
    verbose=1
)
```

```python
from rdkit import Chem
import torch
from torch_geometric.data import Data

def mol_to_graph(smiles):

    mol = Chem.MolFromSmiles(smiles)

    if mol is None:
        return None

    x = []

    for atom in mol.GetAtoms():

        x.append([
            atom.GetAtomicNum(),
            atom.GetDegree(),
            atom.GetFormalCharge(),
            int(atom.GetIsAromatic()),
            atom.GetTotalNumHs()
        ])

    edge_index = []

    for bond in mol.GetBonds():

        start = bond.GetBeginAtomIdx()
        end = bond.GetEndAtomIdx()

        edge_index.append([start, end])
        edge_index.append([end, start])

    x = torch.tensor(
        x,
        dtype=torch.float
    )

    edge_index = torch.tensor(
        edge_index,
        dtype=torch.long
    ).t().contiguous()

    return Data(
        x=x,
        edge_index=edge_index
    )
```

```python
from torch_geometric.data import Batch

def collate_fn(batch):

    cells = []
    graphs = []
    targets = []

    for cell, graph, target in batch:

        cells.append(cell)
        graphs.append(graph)
        targets.append(target)

    cells = torch.stack(cells)

    targets = torch.stack(targets)

    graph_batch = Batch.from_data_list(
        graphs
    )

    return (
        cells,
        graph_batch,
        targets
    )
```

```python
import torch
import torch.nn as nn

from torch_geometric.nn import (
    GCNConv,
    global_mean_pool
)

class DrugGNN(nn.Module):

    def __init__(self):

        super().__init__()

        self.conv1 = GCNConv(
            5,
            64
        )

        self.conv2 = GCNConv(
            64,
            128
        )

        self.fc = nn.Linear(
            128,
            256
        )

    def forward(
        self,
        x,
        edge_index,
        batch
    ):

        x = self.conv1(
            x,
            edge_index
        )

        x = torch.relu(x)

        x = self.conv2(
            x,
            edge_index
        )

        x = torch.relu(x)

        x = global_mean_pool(
            x,
            batch
        )

        x = self.fc(x)

        return x


class DrugResponseGNN(nn.Module):

    def __init__(self):

        super().__init__()

        self.drug_gnn = DrugGNN()

        self.cell_fc = nn.Linear(
            272,
            256
        )

        self.fc1 = nn.Linear(
            512,
            512
        )

        self.fc2 = nn.Linear(
            512,
            256
        )

        self.fc3 = nn.Linear(
            256,
            128
        )

        self.out = nn.Linear(
            128,
            1
        )

        self.dropout = nn.Dropout(
            0.3
        )

    def forward(
        self,
        cell_features,
        graph_batch
    ):

        drug_embedding = self.drug_gnn(
            graph_batch.x.float(),
            graph_batch.edge_index,
            graph_batch.batch
        )

        cell_embedding = torch.relu(
            self.cell_fc(
                cell_features
            )
        )

        x = torch.cat(
            [
                cell_embedding,
                drug_embedding
            ],
            dim=1
        )

        x = torch.relu(
            self.fc1(x)
        )

        x = self.dropout(x)

        x = torch.relu(
            self.fc2(x)
        )

        x = self.dropout(x)

        x = torch.relu(
            self.fc3(x)
        )

        x = self.out(x)

        return x.squeeze()
```

```python
device = torch.device(
    "cuda" if torch.cuda.is_available()
    else "cpu"
)

model = DrugResponseGNN().to(
    device
)

criterion = nn.MSELoss()

optimizer = torch.optim.Adam(
    model.parameters(),
    lr=1e-4
)

print(device)
```

```python
from tqdm import tqdm

num_epochs = 30

train_losses = []

for epoch in range(num_epochs):

    model.train()

    epoch_loss = 0

    loop = tqdm(
        train_loader,
        leave=False
    )

    for cells, graphs, targets in loop:

        cells = cells.to(device)

        graphs = graphs.to(device)

        targets = targets.to(device)

        optimizer.zero_grad()

        preds = model(
            cells,
            graphs
        )

        loss = criterion(
            preds,
            targets
        )

        loss.backward()

        optimizer.step()

        epoch_loss += loss.item()

    avg_loss = (
        epoch_loss /
        len(train_loader)
    )

    train_losses.append(
        avg_loss
    )

    print(
        f"Epoch {epoch+1}/{num_epochs} "
        f"Loss: {avg_loss:.4f}"
    )
```
